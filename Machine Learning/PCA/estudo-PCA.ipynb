# **PCA - Principal Component Analysis**

Esta análise aplica PCA para reduzir a dimensionalidade dos dados, utilizando a base `fetal_health`.

 A base contém 2126 registros extraídos de exames de Cardiotocograma (CTG), que são essenciais para avaliar a saúde fetal. Esses exames foram classificados em três categorias: Normal, Suspeito e Patológico.

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
base = pd.read_csv('../../Base de Dados/fetal_health.csv')
```

```python
base
```

No PCA (Análise de Componentes Principais), a transformação é aplicada apenas nas variáveis preditoras (variáveis explicativas ou X). O objetivo é reduzir a dimensionalidade dos dados, mantendo o máximo de variabilidade possível nas componentes principais.

```python
X = base.drop(columns=['fetal_health'])
```

# Relembrando PCA:

As componentes principais (ou PCs) no PCA são combinações lineares das variáveis originais que capturam a maior parte da variabilidade dos dados. Em vez de observar cada variável individualmente, o PCA cria novas "variáveis" (componentes) que representam os dados de forma mais compacta, concentrando o máximo de informação possível em menos dimensões. As primeiras componentes explicam a maior parte da variação nos dados, permitindo trabalhar com menos variáveis sem perda significativa de informação.

A base possui 21 variáveis, e a redução inicial considera 10 componentes.

Em geral, o objetivo do PCA é reduzir a dimensionalidade ao mesmo tempo em que se preserva a maior parte possível da informação (variabilidade) original.

```python
pca = PCA(n_components= 10 ) # Substitua pelo número de componentes que deseja
# Ajustando o PCA aos dados X
pca.fit(X)
```

```python
# Variância explicada por cada componente
explained_variance = pca.explained_variance_ratio_

# Variância explicada acumulada
cumulative_variance = explained_variance.cumsum()
```

```python
# Plotando a variância explicada acumulada
plt.plot(range(1, 11), cumulative_variance, marker='o', linestyle='--')
plt.title('Variância Explicada Acumulada por Componente')
plt.xlabel('Número de Componentes')
plt.ylabel('Variância Explicada Acumulada')
plt.grid(True)
plt.show()
```

**Interpretando esse gráfico:**

Eixo X (Número de Componentes): representa o número de componentes principais considerados, de 1 até o número de componentes especificado (neste caso, 10).

Eixo Y (Variância Explicada Acumulada): Representa a proporção acumulada da variância total dos dados que é explicada pelas componentes principais.

Curva: mostra como a variância explicada se acumula conforme mais componentes principais são adicionadas.

Observe o "joelho" da curva: é o ponto em que a adição de mais componentes deixa de gerar aumento significativo na variância explicada. Esse ponto indica quantas componentes principais são suficientes. Neste caso, o "joelho" ocorre antes da 10ª componente, sugerindo o uso de menos componentes.

Com 4 componentes, cerca de 90% da variabilidade dos dados originais é capturada. Isso indica que a maior parte da informação relevante pode ser representada por essas 4 componentes.

```python
print(pca.explained_variance_ratio_)
print(pca.singular_values_)
```

O primeiro array mostra a proporção da variância total dos dados explicada por cada componente principal.

O segundo array contém os valores singulares associados a cada componente principal, que representam a magnitude da variância explicada por cada componente.
Esses valores indicam a "força" de cada componente no modelo.

Avaliação da primeira componente:

Primeira Componente (0.5889): A primeira componente principal captura aproximadamente 58,9% da variância total dos dados. Isso significa que quase 60% da informação original está concentrada nessa única componente.


Primeira Componente (2513.27): Este valor singular elevado reflete a grande quantidade de variância capturada pela primeira componente.


**PCA recalculado com apenas 4 componentes:**

```python
pca_4 = PCA(n_components= 4 )
pca_4.fit(X)
```

```python
explained_variance = pca_4.explained_variance_ratio_
cumulative_variance = explained_variance.cumsum()
```

```python
plt.plot(range(1, 5), cumulative_variance, marker='o', linestyle='--') ##ALTERAR O RANGE
plt.title('Variância Explicada Acumulada por Componente')
plt.xlabel('Número de Componentes')
plt.ylabel('Variância Explicada Acumulada')
plt.grid(True)
plt.show()
```

# Contrução do modelo aplicando PCA:

```python
Y = base['fetal_health']
```

```python
X_pca = pca.fit_transform(X)
```

```python
X_pca
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X_pca, Y, test_size=0.3, random_state=42)
```

```python
rf_model = RandomForestClassifier(random_state=42)
```

```python
rf_model.fit(X_train, Y_train)
```

```python
Y_pred = rf_model.predict(X_test)
```

```python
print("Matriz de Confusão:")
print(confusion_matrix(Y_test, Y_pred))

print("\nRelatório de Classificação:")
print(classification_report(Y_test, Y_pred))
```
